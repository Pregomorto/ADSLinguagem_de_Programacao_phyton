```python
#Finalidade de demoonstrar o usos de While independete para repetição focada em validação de dados
print("---CADASTRO NO SISTEMA---")

#VALIDAÇÃO DE IDADE
#Verifica se idade esta dentro do permitido
while True:
    idade=int(input("Digite Idade (Apenas +18!): "))
    if 18 <= idade <= 100:
        break
    else:
        print("\nIdade Não-Permitida! Tente novamente.\n")
#Validação do genero
# O .upper() converte o texto para maiúsculas (ex: 'm' vira 'M')
# Verifica se o valor digitado é uma das opções válidas
while True:
    genero = input("Qual seu Gênero? (M: Masculino, F: Feminino, O: Outro): ").upper()
    if genero in ['M', 'F', 'O']:
        break
    else:
        print(" Gênero Invalido!")
# 3. Confirmação do Cadastro
print(f" Cadastro salvo! Obrigado!")
print(f" Dados Atribuidos: Idade {idade} anos| Gênero {genero}")
```
