```python
#Funções Definidas pelo usuário
def calcular_desconto(valor, percentual):
  """Calcula o valor final aplicando um desconto percentual"""
  if percentual < 0 or percentual > 100:
    return None #marcação invalidade
  desconto = valor * (percentual/100)
  return valor - desconto

def registrar_venda(produto, valor_final):
  """Imprime Registro de uma venda."""
  print(f">>Venda registrada: {produto} por R$ {valor_final:2f}")

  #funcao lambda para arredondar

  arredondar = lambda v: round(v,2)

  #variaveis do PDV
  total_venda = 0
  qtd_venda = 0 # Corrected variable name from qtd_vendar

  #loop principal do caixa
while True:
  print("+++++Minha Loja++++")
  print("1 - Registrar Venda")
  print("2 - Ver relatorio do Dia")
  print("3 - Sair")
  opcao = input("Escolha uma opção: ")

  if opcao == '0':
    print('Encerrando caixa... Até breve!')
    break

  elif opcao == "1":
    produto = input('Nome do Produto: ')
    valor = float(input('Valor do Produto: '))
    percentual = float(input('Percentual de Desconto: '))

    valor_final = calcular_desconto(valor, percentual)

    if valor_final is None:
      print("Desconto Invalido: Use valor entre 0 e 100")
    else:
      valor_final = arredondar(valor_final)
      registrar_venda(produto, valor_final)
      total_venda += valor_final
      qtd_venda += 1

  elif opcao == "2":
    print(f"Vendas hoje: {qtd_venda}")
    print(f"Total de vendas: R$ {arredondar(total_venda)}")

  else:
    print("Opção Inválida")
    if qtd_venda > 0:
      media_venda = total_venda / qtd_venda
```
