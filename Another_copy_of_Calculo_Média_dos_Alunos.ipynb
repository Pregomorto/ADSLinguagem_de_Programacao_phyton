```python
# ==== FUNÇÕES ====
#Resultado final em tese...

#Parte de Cadastro
def cadastrar_alunos():
    alunos = []
    while True:
        nome = input("Nome do aluno (ou 'sair'): ")
        if nome.lower() == 'sair':
            break

        notas = []
        for i in range(7):
            entrada = input(f"Nota {i+1} (ou 'sair'): ")
            if entrada.lower() == 'sair':
                break
            nota = float(entrada)
            notas.append(nota)

        aluno = {"nome": nome, "notas": notas}
        alunos.append(aluno)

    return alunos

#Parte de Relatorios e Caclulos
def exibir_relatorio(alunos):
    print("\n=== RELATÓRIO FINAL ===")
    for aluno in alunos:
        print(f"\nAluno: {aluno['nome']}")

        # ← aqui entra o trecho do gabarito
        notas = aluno['notas']

        if notas:
            soma_notas = 0
            for nota in notas:
                soma_notas += nota
            media = soma_notas / len(notas)
        else:
            media = 0

        if media >= 7:
            situacao = "Aprovado"
        else:
            situacao = "Reprovado"

        print("Notas:", notas)
        print("Média:", media)
        print("Situação:", situacao)


# ++++ PROGRAMA PRINCIPAL +++++++
print("=== GESTOR DE NOTAS ===\n")

alunos = cadastrar_alunos()
exibir_relatorio(alunos)
```
