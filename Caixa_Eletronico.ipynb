```python
saldo_total = 0.0

total_depositos=0

print("\n----CAIXA ELETRÔNICO: DEPÓSITOS ---\n")
print("\nDigite o Valor do Depósito ou pressione '0' para encerrar.\n")

while True:
    valor=float(input("Digite o valor á ser depositado...:"))
    if valor <=0:
        print("\nEncerrando Operações... Até a Próxima!\n")
        break

    saldo_total += valor
    total_depositos += 1
    print(f" Depósito Realizado com Sucesso!...\n")
    print(f" Saldo atual: R$ {saldo_total:.2f}.\n")

print("\n---RELATÓRIO DE DADOS E OPERAÇÕES---\n")
print(f" Total de Depósito e Operações Realizadas: {total_depositos}")
print(f" saldo total Acumulado: R$ {saldo_total:.2f}")
```
