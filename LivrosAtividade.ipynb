```python
# ==== IMPORTAÇÃO do mat ====
import matplotlib.pyplot as plt


# ==== PASSO 1: CLASSE LIVRO ====
class Livro:
    # __init__ prepara o livro/conteud
    def __init__(self, titulo, autor, genero, quantidade):
        self.titulo = titulo          # O"eu mesmo" guardo o título
        self.autor = autor            # "eu mesmo" guardo o autor
        self.genero = genero          # "eu mesmo" guardo o gênero
        self.quantidade = quantidade  # "eu mesmo" guardo a quantidade

    # para mostrar o livro
    def __str__(self):
        return f"{self.titulo} - {self.autor} ({self.genero}) - {self.quantidade} unid."


# ==== PASSO 2: LISTA DE LIVROS ====
biblioteca = []


# ==== PASSO 3: FUNÇÕES ====

# Função 1: Cadastra livroex
def cadastrar_livro(titulo, autor, genero, quantidade):
    novo_livro = Livro(titulo, autor, genero, quantidade)
    biblioteca.append(novo_livro)
    print(f" Livro '{titulo}' cadastrado!")

# Função 2: Dispor todos os livros
def listar_livros():
    if not biblioteca:
        print(" Nenhum livro cadastrado.\n")
        return

    print("\n LISTA DE LIVROS:")
    for i, livro in enumerate(biblioteca, 1):
        print(f"{i}. {livro}")

# Função 3: Busca livro pelo titu
def buscar_livro(titulo):
    for livro in biblioteca:
        if livro.titulo.lower() == titulo.lower():
            print(f"🔍 Encontrado: {livro}")
            return
    print(f"❌ Livro '{titulo}' não encontrado.")


# ==== PASSO 4: GRÁFICO ====
def gerar_grafico():
    from collections import Counter

    if not biblioteca:
        print("📭 Nenhum livro para gerar gráfico.")
        return

    # Conta generos dos livros
    generos = [livro.genero for livro in biblioteca]
    contagem = Counter(generos)

    # Cria o gráfico dos livros
    plt.figure(figsize=(8, 5))
    plt.bar(contagem.keys(), contagem.values(), color=['#3498db', '#e74c3c', '#2ecc71', '#f39c12'])
    plt.title('Quantidade de Livros por Gênero')
    plt.xlabel('Gênero')
    plt.ylabel('Quantidade')
    plt.show()


# ==== PASSO 5: TESTAR O SISTEMA ====
print("=== SISTEMA DE BIBLIOTECA ===\n")

# Cadastrando livros
cadastrar_livro("Antes do baile verde ", " Lygia Fagundes Telles", " Ficção", 5)
cadastrar_livro("Quarto de Despejo", "Carolina Maria de Jesus", "Auto Biografia", 3)
cadastrar_livro("O Poder da Comunicação", "Osmar Barbosa", "Comunicação", 7)
cadastrar_livro("Comunicação Plural Em Cena", "ECA-USP", "Artígo Acadêmico", 4)

# Listando
listar_livros()

# Buscando
buscar_livro("Antes do Baile Verde")

# Gráfico
gerar_grafico()
```

```python
# ==== IMPORTAÇÃO ====
import matplotlib.pyplot as plt


# ==== PASSO 1: CLASSE LIVRO ====
class Livro:
    # __init__ prepara o livro/conteúdo
    def __init__(self, titulo, autor, genero, quantidade):
        self.titulo = titulo          # "eu mesmo" guardo o título
        self.autor = autor            # "eu mesmo" guardo o autor
        self.genero = genero          # "eu mesmo" guardo o gênero
        self.quantidade = quantidade  # "eu mesmo" guardo a quantidade

    # para mostrar o livro
    def __str__(self):
        return f"{self.titulo} - {self.autor} ({self.genero}) - {self.quantidade} unid."


# ==== PASSO 2: LISTA DE LIVROS ====
biblioteca = []


# ==== PASSO 3: FUNÇÕES ====

# Função 1: Cadastra livro
def cadastrar_livro(titulo, autor, genero, quantidade):
    # .strip() remove espaços sobrando antes de salvar
    novo_livro = Livro(titulo.strip(), autor.strip(), genero.strip(), quantidade)
    biblioteca.append(novo_livro)
    print(f"✅ Livro '{novo_livro.titulo}' cadastrado!")


# Função 2: Listar todos os livros
def listar_livros():
    if not biblioteca:
        print("📭 Nenhum livro cadastrado.\n")
        return

    print("\n📚 LISTA DE LIVROS:")
    for i, livro in enumerate(biblioteca, 1):
        print(f"{i}. {livro}")


# Função 3: Busca livro pelo título
def buscar_livro(titulo):
    for livro in biblioteca:
        # emove espaços E ignora maiúsculas/minúsculas
        if livro.titulo.strip().lower() == titulo.strip().lower():
            print(f"🔍 Encontrado: {livro}")
            return
    print(f"❌ Livro '{titulo}' não encontrado.")


# ==== PASSO 4: GRÁFICO ====
def gerar_grafico():
    from collections import Counter

    if not biblioteca:
        print("📭 Nenhum livro para gerar gráfico.")
        return

    # Conta gêneros dos livros
    generos = [livro.genero for livro in biblioteca]
    contagem = Counter(generos)

    # Cria o gráfico dos livros
    plt.figure(figsize=(8, 5))
    plt.bar(contagem.keys(), contagem.values(), color=['#3498db', '#e74c3c', '#2ecc71', '#f39c12'])
    plt.title('Quantidade de Livros por Gênero')
    plt.xlabel('Gênero')
    plt.ylabel('Quantidade')
    plt.show()


# ==== PASSO 5: TESTAR O SISTEMA ====
print("=== SISTEMA DE BIBLIOTECA ===\n")

# Cadastrando livros
cadastrar_livro("Antes do Baile Verde", "Lygia Fagundes Telles", "Ficção", 5)
cadastrar_livro("Quarto de Despejo", "Carolina Maria de Jesus", "Auto Biografia", 3)
cadastrar_livro("O Poder da Comunicação", "Osmar Barbosa", "Comunicação", 7)
cadastrar_livro("Comunicação Plural Em Cena", "ECA-USP", "Artigo Acadêmico", 4)

# Listando
listar_livros()

# Buscando (testando variações de maiúsculas/minúsculas)
buscar_livro("Antes do Baile Verde")     # com maiúsculas
buscar_livro("antes do baile verde")     # tudo minúsculo
buscar_livro("ANTES DO BAILE VERDE")     # tudo maiúsculo
buscar_livro("  Antes do Baile Verde  ") # com espaços sobrando

# Gráfico
gerar_grafico()
```
